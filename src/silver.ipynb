```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.events AS

SELECT
    event_id,
    user_id,
    product_id,
    event_type,
    TO_DATE(event_timestamp, 'yyyy-MM-dd') AS event_timestamp
FROM datacleaning.bronze.events;
```

```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.order_items AS

SELECT
    order_item_id,
    order_id,
    product_id,	
    user_id,	
    quantity,	
    item_price,	
    item_total
FROM datacleaning.bronze.order_items;
```

```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.orders AS

SELECT
    order_id,	
    user_id,
    order_date,	
    order_status,	
    total_amount
FROM datacleaning.bronze.orders;
```

```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.products AS

SELECT
    product_id,	
    product_name,	
    category,	
    brand,	
    price,	
    rating
FROM datacleaning.bronze.products;
```

```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.reviews AS

SELECT
    review_id,	
    order_id,	
    product_id,	
    user_id	rating,	
    review_text,	
    review_date
FROM datacleaning.bronze.reviews;
```

```python
%sql
CREATE OR REPLACE TABLE datacleaning.silver.users AS

SELECT
    user_id	name,	
    email,	
    gender,	
    city,	
    signup_date
FROM datacleaning.bronze.users;
```
